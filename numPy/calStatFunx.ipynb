#Calculus and statistical functions

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a1 = 2 * np.random.randn(10000) + 10
np.mean(a1) #mean of the normal distribution
np.std(a1) #standard deviation of the normal distribution
```

```python
np.percentile(a1,99) #gives the number for which 80% of the numbers are less than in the array
```

```python
x = np.linspace(1,10,100)
y = 1/x**2 * np.sin(x)
dydx = np.gradient(y,x)
yint = np.cumsum(y) * (x[1] - x[0])
plt.plot(x,y)
plt.plot(x,dydx)
```
