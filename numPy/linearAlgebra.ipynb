# Basic Linear Algebra

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
A = np.array([[3,2,1],[5,-5,4],[6,0,1]]) #matrices
b1 = np.array([1,2,3]) #vector
b2 = np.array([-1,2,-5])

A
```

```python
A@b1 #multiply matrci with a vector
```

```python
A.T #transpose of the matrix
```

```python
np.dot(b1,b2) #dot product of two vectors
```

```python
np.cross(b1,b2) #cross product
```

```python
#Applying Ax = b

#defined A above

b = np.array([4,3,0])
```

```python
np.linalg.solve(A,b)
```

```python
A = np.array([[4,2,2],[2,4,2],[2,2,4]])

w,v = np.linalg.eig(A) # destructuring eigen values and eigen values

v1 = v[:,0] #first eigen vector
v1
```

```python
A@v1
```

```python
w[0] * v1
```
