# Array Indexing and slicing

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a1 = np.array([1,4,3,6,9])
```

```python
a1[2:] # gives elements above index 2
a1[:-2] # give the elemtns from 0 to 2, -2 indicates the end point here

a1[1:-2] #gives elements from index 1 to the 2nd index

# for boolean operations

a1>4 # gives true false array
a1[a1>4] # gives elements of a1 with true a1
```

```python
names  = np.array(['James','Jim','Pete','Cramar'])
first_letter_j = np.vectorize(lambda s: s[0])(names) == 'J'
names[first_letter_j]
```

```python
a1[a1%4 == 0]
```
