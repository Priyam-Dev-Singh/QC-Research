```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,10,10001)
def f(x):
    return np.exp(-x/10)* np.sin(x)
y = f(x)

plt.plot(x,y)
```

```python
p = y[(x>=4)*(x<=7)]
p.mean() #-0.2956023068963138
p.std() #0.31448753875085117
```

```python
np.percentile(p,80)
```

```python
dydx = np.gradient(y,x)
plt.plot(x,dydx)
```

```python
x[1:][dydx[1:] * dydx[:-1]<0]
```
