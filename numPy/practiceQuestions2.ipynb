```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-2,2,1000)
y = np.linspace(2,-2,1000)

xv,yv = np.meshgrid(x,y)
f = np.exp(-(xv**2 + yv**2)) * np.sin(xv)
plt.contourf(xv,yv, f, levels = 100)
plt.colorbar()
```

```python
np.abs(f.ravel()).sum() * np.diff(x)[0] * np.diff(y)[0]
```

```python
np.abs(f[xv ** 2 + yv**2 > 0.5 **2]
.ravel()).sum() * np.diff(x)[0] * np.diff(y)[0]
```

```python
A = np.array([[3,2,3,10],[2,-2,5,8],[3,3,4,9],[3,4,-3,-7]])
b = np.array([4,1,3,2])
np.linalg.solve(A,b)
```
