# 2D Plots

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
_ = np.linspace(-1,1,100)
x,y = np.meshgrid(_,_)
z = x**2 + x*y
```

Filled in contour plots

```python
plt.contourf(x,y,z, levels = 30, vmin= 0.72, vmax = 1.68)
plt.colorbar(label='ABC')
#vmin and vmax sets a range for max and min for the colors
```

Contour Plot

```python
cs = plt.contour(x,y,z, levels = 30)
plt.clabel(cs ,fontsize = 8)
```

3D Surface plots

```python
fig, ax = plt.subplots(subplot_kw={'projection' : '3d'})
ax.plot_surface(x,y,z, cmap = 'coolwarm') # to draw the contour on the 3d plane
```

Stream Plots: For plotting vector fields

```python
w = 3
_ = np.linspace(-3,3,100)
X,Y = np.meshgrid(_,_)
U = -1 - X**2 +Y
V = 1 + X - Y**2
speed = np.sqrt(U**2 + V**2)
```

```python
fig, axes = plt.subplots(2,2, figsize = (7,7))

ax = axes[0][0]
ax.streamplot(X,Y,U,V)
ax = axes[0][1]
ax.streamplot(X,Y,U,V, color=speed)
ax = axes[1][0]
lw = 5*speed / speed.max()
ax.streamplot(X,Y,U,V, linewidth=lw) # we can use it to demostrate the density of the effect more visually appealing
ax = axes[1][1]
lw = 5*speed / speed.max()
seedpoints = np.array([[0,1],[1,0]])
ax.streamplot(X,Y,U,V, start_points= seedpoints) #to see the images corresponding to distinct lines or points
```
