# Advanced API

```python
import numpy as np
import matplotlib.pyplot as plt
```

Single Axes in one figure

```python
fig, ax = plt.subplots(1,1, figsize = (12,4))
x = np.linspace(0, 15, 30)
y = np.sin(x) + 0.1*np.random.randn(len(x))
plt.plot(x,y, 'o--')
ax.set_xlabel('adsc')
ax.set_ylabel('aagsrbsdf')
```

```python
fig, axes = plt.subplots(3,2, figsize = (12,12))
ax = axes[0][0]
x = np.linspace(0, 15, 30)
y = np.sin(x) + 0.1*np.random.randn(len(x))
ax.set_xlabel('fadb')
ax.set_ylabel('oihfoasbl')
ax.tick_params(axis = 'both', labelsize = 10)
ax.text(0.1,0.1, 'text-here', transform = ax.transAxes)
ax.plot(x,y, 'o--')

#here we are to select the coordinated of the graph we want to chose to construct our graph adn then we have to build the graph using the indexed axes of the axes array instead of chossing plt, this tells the interpretor that go to ax and plot this 
ax = axes[0][1]
res = np.random.randn(1000)*0.4 + 0.9
ax.hist(res, density= True, bins=30,label='op',histtype='step')
ax.legend()
ax.set_title('Histogram')
plt.show()
```
