```python
import numpy as np
import matplotlib.pyplot as plt

```

Line Plots

```python
x = np.linspace(0, 15, 30)
y = np.sin(x) + 0.1*np.random.randn(len(x))
plt.plot(x,y)
```

```python
plt.plot(x,y, 'o--', color = 'purple', lw = 0.7, ms = 5)
```

```python
plt.figure(figsize=(10,3)) #size of the graph image length, breadth
plt.plot(x,y, 'o--', color = 'purple', lw = 0.7, ms = 5)
plt.xlabel('Time [s]') #adding x and y labels
plt.ylabel('Voltage[V]')
```

```python
plt.figure(figsize=(10,3)) #size of the graph image length, breadth
plt.plot(x,y, 'o--', color = 'purple', lw = 0.7, ms = 5, label = 'Component 1')
plt.xlabel('Time [s]') #adding x and y labels
plt.ylabel('Voltage[V]')
plt.legend(loc = 'upper right', fontsize = 10)
```

```python
#Multiple lines on the same plot

x2 = np.linspace(0,15,100)
y2 = np.sin(x2)
```

```python
plt.figure(figsize = (8,3))
plt.plot(x2,y2, lw = 2, ms = 6, color='orange', label ='sin(x)')
plt.plot(x,y, 'o', color = 'purple', lw = 0.7, ms = 5, label = 'Component 1')
plt.xlabel('Time')
plt.ylabel('Voltage')
plt.legend(loc = 'upper right', fontsize = 10)
```

```python
plt.figure(figsize = (8,3))
plt.plot(x2,y2, lw = 2, ms = 6, color='orange', label ='sin(x)')
plt.plot(x,y, 'o', color = 'purple', lw = 0.7, ms = 5, label = 'Component 1')
plt.xlabel('Time')
plt.ylabel('Voltage')
plt.ylim(top = 2.5)
plt.legend(loc = 'upper right', fontsize = 10, ncol = 2)
```
