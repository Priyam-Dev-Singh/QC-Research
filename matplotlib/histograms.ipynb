## Histograms

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
res = np.random.randn(1000)*0.2 + 0.4
```

```python
a = plt.hist(res)
```

```python
a
```

```python
plt.figure(figsize=(8,3))
plt.hist(res, bins = 30, density=True) #densire true makes the volume of the histogram 1
plt.show()
```

```python
res2 = np.random.randn(1000)*0.2 + 0.4
```

```python
plt.figure(figsize=(8,3))
plt.hist(res, bins = 30, density=True, histtype='step')
plt.hist(res2, bins = 30, density=True, histtype='step')
plt.show()
```
