# Calculus

## Differentiation

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
from scipy.differentiate import derivative
```

```python


def f(x):
    return x**2 * np.sin(2*x)*np.exp(-x)

x= np.linspace(0,1,100)
```

```python
der = derivative(f, x, initial_step = 1e-6, order=2)
```

```python
plt.plot(x, f(x))
plt.plot(x, der.df)
```

# integration

```python
from scipy.integrate import quad
```

```python
integrand = lambda x : x**2 * np.sin(x) * np.exp(-x)
integral, inegral_error = quad(integrand, 0, 1)
```

```python
integral
```

```python
inegral_error
```

## Double integration

```python
from scipy.integrate import dblquad
```

```python
integrand = lambda x, y: np.sin(x+y**2)
lwr_y = lambda x: -x
upr_y = lambda x: x**2
integral, integral_error = dblquad(integrand, 0, 1, lwr_y, upr_y )
```

```python
integral
```

```python
integral_error
```
