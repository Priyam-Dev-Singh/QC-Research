# Statistics

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

Beta distributions

```python
from scipy.stats import beta
```

Basic Stats

```python
a,b = 2.5, 3.1
mean, var, skew, kurt = beta.stats(a, b, moments = 'mvsk')
```

```python
var
```

Plotting Probability Distribution

```python
beta.ppf(0.99, a,b)
```

```python
x = np.linspace(beta.ppf(0, a,b), beta.ppf(1, a,b), 100)
plt.plot(x, beta.pdf(x,a,b))
```

Guassian Distributions

```python
from scipy.stats import norm
```

```python
mu = 1
sigma = 2
mean, var = norm.stats(loc = mu, scale = sigma, moments = 'mv')
```

```python
mean
```

```python
x = np.linspace(norm.ppf(0.01, mu,sigma), norm.ppf(0.99, mu,sigma), 100)
plt.plot(x, norm.pdf(x,mu,sigma))
```

Multinomial Distributions

```python
from scipy.stats import multinomial
```

```python
p = np.ones(6)/6  #this tell us that what is the probability if we roll the six times (n=6) and get 6 1s (the array) for the prob dist p
multinomial.pmf([6,0,0,0,0,0], n=6, p=p)
```

```python
# We can also generate random multinomial PD here we are conducting each experiment 100 events of rolling a dice for prob dist p and we do each experiment 5 times so we get 2 D array of times each number came on dice
multinomial.rvs(n=100, p=p, size=5)
```
