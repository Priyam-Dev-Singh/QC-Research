# Basic Linear Algebra Using Scipys

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
```

```python
#sci py provides specific matrices solver to proovide more easable use

#below is used to oslve triangular matrices
from scipy.linalg import solve_triangular
```

```python
a = np.array([[3,0,0,0],[2,1,0,0],[1,0,1,0],[1,1,1,1]])
b = np.array([4,2,4,2])

x = solve_triangular(a,b,lower=True)
x
```

```python
s = np.linalg.solve(a,b)
s
```

We can also solve below toeplitz matrix (matrices with constant diagonal) using pre built functions of scipy

```python
from scipy.linalg import solve_toeplitz, toeplitz
```

```python
c = np.array([1,3,6,10]) #first column of T
r = np.array([1,-1,2,3]) #first row of T
b = np.array([1,2,2,5])
x = solve_toeplitz((c,r),b)
```

```python
x
```

### Eigen Value Problems

```python
#above matrix is tridiagonal as well as toeplitz

from scipy.linalg import eigh_tridiagonal
```

```python
d = 3*np.ones(4)
e = -1*np.ones(3)
w,v = eigh_tridiagonal(d,e)
```

```python
v.T[1]
```

```python
A = np.diag(d) + np.diag(e, k=1) + np.diag(e, k=-1)
A
```

```python
#to check if our eigen vectors are correct

A@v.T[1]
```

```python
w[1]*v.T[1]
```
