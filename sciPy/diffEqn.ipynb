#### Coupled first order ODEs


```python
def dSdx(S,x):
    y1,y2  =S
    return [ y1+y2**2 + 3*x, 3*y1+y2**3 - np.cos(x) ]

y1_0 = 0
y2_0 = 0
S_0 = (y1_0, y2_0 )
```

```python
x = np.linspace(0,1,100)
sol = odeint(dSdx, S_0, x)
```

```python
y1 = sol.T[0]
y2 = sol.T[1]
```

```python
plt.plot(x,y1)
plt.plot(x, y2)
```

Any second order ODE can be transformed into two first order ODEs to solve the differential equation as it cant be solves directly

Here, we convert the equation of an pundulum:

$$ \theta'' - \sin(\theta)$$

can we converted to a system of two coupled DE

$$ d\omega/dt = \sin(\theta) $$
$$ d\theta/dt = \omega $$

Let $$ S = (\theta, \omega)$$

```python
def dSdt(S,t):
    theta, omega = S
    return [omega, np.sin(theta)]

theta0 = np.pi/4
omega0 = 0
S0 = (theta0, omega0)
```

```python
t = np.linspace(0,20,100)
sol = odeint(dSdt, S0, t)
theta, omega = sol.T
```
